### Exercise 5(b): Median

Here I calculate the median (middle value) manually by sorting the data and selecting the middle
element(s). Then I confirm the result using the `statistics` module.

```python
# Sort the salaries so I can locate the middle value(s)
sorted_salaries = sorted(salaries)

# Store the count of observations
n = len(sorted_salaries)

# Manual median calculation:
# - if n is even: average the two middle values
# - if n is odd: take the single middle value
if n % 2 == 0:
    median_manual = (sorted_salaries[n//2 - 1] + sorted_salaries[n//2]) / 2
else:
    median_manual = sorted_salaries[n//2]

median_manual
```

```python
# Median using the statistics module
st.median(salaries)
```

### Exercise 5(d): Sample variance

Here I compute the sample variance manually using the sample variance formula (divide by n-1),
then compare it to the result from `statistics.variance()`.

```python
# Manual sample variance calculation (using the sample variance formula)

# Compute the mean using basic Python only
mean_manual = sum(salaries) / len(salaries)

# Store the number of observations
n = len(salaries)

# Manual sample variance:
# 1) compute squared deviations from the mean
# 2) divide by (n - 1) for sample variance
variance_manual = sum((x - mean_manual) ** 2 for x in salaries) / (n - 1)

variance_manual
```

```python
# Sample variance using the statistics module
# This should match the manual result from the previous cell
st.variance(salaries)
```

### Exercise 5(e): Sample standard deviation

Here I compute the sample standard deviation manually as the square root of the sample variance,
then confirm the result using `statistics.stdev()`.

```python
# Manual standard deviation: square root of the manual sample variance
std_manual = variance_manual ** 0.5
std_manual
```

```python
# Sample standard deviation using the statistics module
st.stdev(salaries)
```

## Chapter 1, Exercise 7: Scaling the data

In this exercise, I scale the salary data in two ways:
1) Min-max scaling (normalization) to convert values into the 0–1 range
2) Standardization to convert values into z-scores (mean 0, standard deviation 1)

Exercise 7(a) Min-max Scaling (Normalizing)

```python
# Min-max scaling:
# scaled = (x - min) / (max - min)
min_val = min(salaries)
max_val = max(salaries)

scaled_salaries = [(x - min_val) / (max_val - min_val) for x in salaries]

# Preview the first few scaled values
scaled_salaries[:10]
```

Exercise 7(b) Standardizing

```python
# Standardization (z-scores):
# standardized = (x - mean) / std_dev
mean_val = st.mean(salaries)
std_val = st.stdev(salaries)

standardized_salary = [(x - mean_val) / std_val for x in salaries]

# Preview the first few standardized values
standardized_salary[:10]
```

Exercise 8(a) The covariance between the standardized and normalized data

In this part, I calculate the covariance between the normalized (min–max scaled) and standardized
salary values. I first compute the covariance using NumPy and then confirm the result using a
manual covariance formula.

```python
# Convert both scaled datasets to NumPy arrays so I can use NumPy covariance functions
scaled_arr = np.array(scaled_salaries)
standardized_arr = np.array(standardized_salary)

# np.cov returns a 2x2 covariance matrix
# The off-diagonal value [0, 1] is the covariance between the two datasets
cov_matrix = np.cov(scaled_arr, standardized_arr)
covariance_value = cov_matrix[0, 1]

# Display the covariance matrix and the covariance between the two datasets
cov_matrix, covariance_value
```

Note: The covariance matrix shows how each dataset varies with itself (diagonal values) and with
the other dataset (off-diagonal values). The off-diagonal value represents the covariance between
the normalized and standardized salaries, which measures how they vary together.

```python
# Manual covariance calculation (since statistics.covariance is not available in this Python version)

# Convert lists to NumPy arrays
x = np.array(scaled_salaries)
y = np.array(standardized_salary)

# Compute the mean of each dataset
mean_x = x.mean()
mean_y = y.mean()

# Manual covariance formula:
# covariance = sum((xi - mean_x)(yi - mean_y)) / (n - 1)
n = len(x)
cov_manual = sum((x[i] - mean_x) * (y[i] - mean_y) for i in range(n)) / (n - 1)

cov_manual
```

Exercise 8(b) The Pearson correlation coefficient between the standardized and normalized data

In this part, I calculate the Pearson correlation coefficient between the normalized and standardized
salary values using NumPy. This measures the strength and direction of the linear relationship between
the two scaled datasets. Because both datasets are linear transformations of the same original data,
the correlation is expected to be very close to 1 and may be exactly 1 depending on rounding.

```python
# Pearson correlation using NumPy
# corrcoef returns a 2x2 correlation matrix; the off-diagonal value is the correlation
corr_matrix = np.corrcoef(scaled_arr, standardized_arr)
correlation_value = corr_matrix[0, 1]

corr_matrix, correlation_value
```

Note. A correlation value very close to 1 indicates a very strong positive linear relationship
between the normalized and standardized datasets. This is expected because both datasets are
different scaled versions of the same original salary data.

## Chapter 2 – Exercise 4: Ring of Fire indicator

In this exercise, I create a new column that indicates whether each earthquake occurred in a
Ring of Fire location. I use the provided list of countries and US states and make sure that
"New Mexico" is not incorrectly matched as "Mexico".

```python
# List of Ring of Fire locations provided in the assignment
ring_locations = [
    "Alaska", "Antarctic", "Bolivia", "California", "Canada", "Chile", "Costa Rica",
    "Ecuador", "Fiji", "Guatemala", "Indonesia", "Japan", "Kermadec", "Mexico",
    "New Zealand", "Peru", "Philippines", "Russia", "Taiwan", "Tonga", "Washington"
]

def is_ring_of_fire(place):
    """
    Return True if the earthquake location is part of the Ring of Fire.
    Includes a safeguard to avoid counting 'New Mexico' as 'Mexico'.
    """
    if not isinstance(place, str):
        return False

    # Avoid incorrectly matching 'New Mexico' as 'Mexico'
    if "New Mexico" in place:
        return False

    # Check each Ring of Fire location
    for loc in ring_locations:
        if loc == "Mexico":
            # Only count Mexico if it is not New Mexico
            if ("Mexico" in place) and ("New Mexico" not in place):
                return True
        else:
            if loc in place:
                return True

    return False

# Create the Ring of Fire indicator column
df["ring_of_fire"] = df["place"].apply(is_ring_of_fire)

# Preview a few rows to confirm the new column was created correctly
df[["place", "ring_of_fire"]].head()
```
